```python
import logging

from ultralytics import YOLO

logging.basicConfig(level=logging.INFO, format="%(asctime)s [%(levelname)s] %(message)s", force=True)
logger = logging.getLogger(__name__)

DATA_YAML_PATH = "/content/phone_yolo_dataset/data.yaml"
BASE_MODEL = "yolov8n.pt"  # nano - fastest, good starting point for a single-class real-time detector;
                            # step up to yolov8s.pt if accuracy matters more than speed once this baseline works
EPOCHS = 50
IMAGE_SIZE = 640
BATCH_SIZE = 16
RUN_NAME = "phone_detector"
```

```python
def main():
    model = YOLO(BASE_MODEL)  # downloads pretrained COCO weights automatically on first run

    logger.info("Starting fine-tuning on %s for %d epochs.", DATA_YAML_PATH, EPOCHS)
    model.train(
        data=DATA_YAML_PATH,
        epochs=EPOCHS,
        imgsz=IMAGE_SIZE,
        batch=BATCH_SIZE,
        name=RUN_NAME,
    )

    logger.info("Running validation...")
    metrics = model.val()
    logger.info("mAP50: %s, mAP50-95: %s", metrics.box.map50, metrics.box.map)

    logger.info(
        "Done. Best weights at runs/detect/%s/weights/best.pt - "
        "point phone_detector.py's MODEL_PATH at that file.",
        RUN_NAME,
    )


if __name__ == "__main__":
    main()
```
